## Statements

if – Do this code ONLY IF the condition is True.

elif (Else If) – Check this ONLY IF the previous condition was False.

else (Otherwise) – Do this IF ALL previous conditions were False.

---------------

in – True IF a value is found inside something (like a list or text).

not in – True IF a value is NOT found inside something.

--------------------------------------------------------------------------------------------------------------------------------------------

and – True ONLY IF BOTH conditions are True.

or – True IF AT LEAST ONE condition is True.

not – Reverses the condition (True becomes False, False becomes True).


SyntaxError: invalid syntax – Grammar mistake in your code.

````python
import numpy as np
import sympy as sp 
fcd, eps_c, eps_c2, n = sp.symbols('fcd eps_c eps_c2 n')

epsilon_c = float(input("Enter the value of epsilon_c: "))

if epsilon_c < 0:
    print("Error: epsilon_c must be non-negative.")
elif epsilon_c > 0 and epsilon_c <= 0.002:
    sigma = fcd * (1 - (1 - eps_c / eps_c2) ** n)
else:
    sigma = fcd

sigma
````

## Lists, Dictionary, Tuple

List [ ] – Ordered collection of items. Can be changed.

```items = ["C30/37", variable]``` to print we can choose which one we can print(items[0])

Dictionary { } – Collection of Key-Value pairs. Used to store properties of an object.

`C30/37 = {"fck": 30, "Ec": 32}` to print we have to indicate name print(C30/37["fck"])

Tuple ( ) – Ordered collection of items. CANNOT be changed (immutable).

`node_1 = (5.0, 2.2)`

### More of Lists

insert(indeks, element): wstawia element na określonej pozycji w liście.

remove(element): usuwa pierwsze wystąpienie elementu w liście.

pop(indeks): usuwa i zwraca element na określonej pozycji w liście.

sort(): sortuje elementy listy w porządku rosnącym.

reverse(): odwraca kolejność elementów w liście.

## LOOPS

### for

`for item in list:`              # Loops through each item in a list

`for key in dict.keys():`       # Gets only the keys (names, e.g., "IPE_220")

`for val in dict.values():`     # Gets only the values (data, numbers)

`for key, val in dict.items():` # Gets BOTH keys and values at the same time

`range(0, 5):`                   # Integers only: 0, 1, 2, 3, 4 (stops before 5)

`range(0,100,2):`               # Integers from 0 to 100 with step "2"

`np.arange(0, 5, 0.5):`          # Decimals with a fixed STEP (e.g., every 0.5)

`np.linspace(0, 10, 5)`          # Exactly 5 POINTS evenly spaced from 0 to 10


````python
# Lists: 

Results = [1 , 2.5 , 9 , 10 ]

new_number = 0

for numbers in Results :
    new_number += numbers
    print(new_number)

````

````python
# Dictionary

Steel_profile = {
    "IPE_300": (300, 150, 7.1, 10.7),
    "IPE_220": (220, 110, 5.9, 9.2), 
 }

# name -> IPE_300
# dimensions -> 300,150,7.1,10.7

for name,dimensions in Steel_profile.items():   # .items() both , .keys() only name, .values() only values
    h,b,tw,tf = dimensions
    print(name,dimensions)
    print(name,f"h:{h}mm",f"b:{b}mm",f"tw:{tw}mm",f"tf:{tf}mm")
````

````python
last_value = 0 

for value in np.linspace(0,100,6):
    if value <= 60:
        last_value = value

print(f"last number below 60 is {last_value}")
````
