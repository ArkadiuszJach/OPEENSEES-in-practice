```python
'Profile: IPE 300 , A = 53.8 cm2 , E = 210 GPa , I = 8357 cm4 , Sw = 42.2 kg/m ' 

L = 10      # m
r = 5       # m
E = 210     # GPa
I = 8357    # cm4
Wel = 557.1 # cm3     Wpl = 628.4

' Material '
fy = 355    # MPa
E = 210     # GPa

' Loads'
gk = 0.46   # kn/m2
sk = 0.56   # kN/m2


'ULS'
q_uls = (gk * 1.35 + sk * 1.5) * r

'SLS'
q_sls = (gk + sk) * r

'Verification'

M_uls = q_uls * L**2 * 0.125 
M_Rd = Wel * fy*1e-3

V_uls = q_uls * L /2 


if M_uls < M_Rd:
    ver_1 = " Cumplied !!! "
else:
    ver_1 = " Failed"


a = 5 / 384 * q_sls * L**4 / (E * I) * 1e5
a_min = L *1e3 / 250


if a < a_min :
    ver_2 = " Cumplied !!! "
else:
    ver_2 = " Failed"


' Results '

print(f"Results: ")
print(f"ULS: M_Ed = {M_uls:.2f}kNm < M_Rd = {M_Rd:.3f} kNm --->{ver_1} {M_uls/M_Rd*100:.2f}%")
print(V_uls)


print(f"SLS: a = {a:.2f}mm < a_min = {a_min:.2f} mm --->{ver_2} {a/a_min*100:.2f}%")


```
